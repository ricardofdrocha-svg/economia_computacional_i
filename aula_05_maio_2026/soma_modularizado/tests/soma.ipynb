```python
def leitura_dados():
    dados = [0,0]
    i=0
    while i < 2:
        dados[i] = float(input(f"\nDigite a parcela {i+1}: "))
        i+=1
    return dados
```

```python
dados = leitura_dados()
```

```python
def soma(x, y):
    return x + y
```

```python
soma(2,3)
```

```python
resultado = soma(dados[0], dados[1])
```

```python
def saida(lista, resultado):
    print(f'\nA soma da parcela {lista[0]} com a parcela {lista[1]} é igual a {resultado}\n')
```

```python
saida(dados, resultado)
```
